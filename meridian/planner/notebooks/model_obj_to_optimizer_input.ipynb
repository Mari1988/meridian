```python
"""Convert Meridian model file to Excel format compatible with FlexibleBudgetPlanner."""
import os
import pandas as pd
import xarray as xr
from meridian.model import model
from meridian import constants as c

from model_to_excel_converter import convert_model_to_excel
```

```python
model_type = 'champ'
```

```python
home_dir = '/Users/mariappan.subramanian/Library/CloudStorage/OneDrive-TheTradeDesk/MMM/BudgetOptimizer'
input_dir = f'{home_dir}/model_objects/{model_type}'
output_dir = f'{home_dir}/optimizer_inputs/{model_type}'
model_files = [file for file in os.listdir(input_dir) if file.endswith('.pkl')]
advertisers = ['Popeyes', 'Allergan', 'Chick-Fil-A', 'MRG_Chevy_LMA', 'Mazda', 'Samsung_US_Starcom', 'Audi', 'Burger_King']
```

```python
model_files
```

```python
for model_file in model_files:
  model_path = f'{input_dir}/{model_file}'
  client = [client for client in advertisers if client in model_file]
  if len(client) == 1:
    output_path = f'{output_dir}/{client[0]}_optimizer_input_{model_type}.xlsx'
    convert_model_to_excel(model_path, output_path)
  else:
    raise ValueError(f'{model_file} is invalid')
```

```python
opt_period = {}
for model_file in model_files:
  model_path = f'{input_dir}/{model_file}'
  client = [client for client in advertisers if client in model_file][0]
  opt_period[client] = {}
  mmm = model.load_mmm(model_path)
  opt_period[client]['start_date'] = mmm.input_data.time[-52:].min().item()
  opt_period[client]['end_date'] = mmm.input_data.time[-52:].max().item()
```
