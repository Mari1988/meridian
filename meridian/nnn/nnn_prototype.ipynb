```python
import numpy as np
```

```python
regions, weeks = 3, 52
channels = 3  # 0: YouTube Ads, 1: Search Ads, 3: Organic Search
embed_dim = 4
np.random.seed(42)

# qualitative features of each channel
channel_embeds = np.random.randn(channels, embed_dim)

# initialize data arrays
X = np.zeros(shape=(regions, weeks, channels, embed_dim))
sales = np.zeros(shape=(regions, weeks))

for r in range(regions):
  media_1 = np.random.rand(weeks) * 10  # YouTube Ads
  media_2 = np.random.rand(weeks) * 5  # Search Ads
  organic = np.zeros(weeks)
  for t in range(weeks):
    organic[t] = 5 + 0.5 * media_1[t] + 0.3 * media_2[t] + (0.1 * organic[t-1] if t > 0 else 0) + np.random.normal(0, 0.1)
    sales[r, t] = 10 + 0.4 * organic[t] + 0.1 * media_1[t] + 0.1 * media_2[t] + np.random.normal(0, 0.5)
  X[r, :, 0, :] = channel_embeds[0] + media_1[:, None]
  X[r, :, 1, :] = channel_embeds[1] + media_2[:, None]
  X[r, :, 2, :] = channel_embeds[2] + organic[:, None]

print(f"X.shape: {X.shape}")
print(f"sales.shape: {sales.shape}")
```
